```python
import os
import duckdb
import polars as pl

# 1. VERIFY PATHS (Adjust if your files are in a subfolder like '/content/public/')
DATA_DIR = "/content/"

train_labels_path = os.path.join(DATA_DIR, "train_labels.csv")
test_path = os.path.join(DATA_DIR, "test.csv")
kyc_path = os.path.join(DATA_DIR, "kyc.parquet")
trx_glob = os.path.join(DATA_DIR, "transactions", "*.parquet")
bal_glob = os.path.join(DATA_DIR, "dayend_balance", "*.parquet")

# 2. INITIALIZE CONNECTION IN THE SAME CELL
print("Initializing DuckDB connection...")
con = duckdb.connect(database=':memory:')

# 3. DEFINE DETAILED AGGREGATION QUERY
print("Extracting high-velocity features via DuckDB...")
query = f"""
WITH base_kyc AS (
    SELECT * FROM read_parquet('{kyc_path}') WHERE ACCOUNT_TYPE = 'Customer'
),
trx_features AS (
    SELECT
        SRC_ACCOUNT as ACCOUNT_ID,
        EPOCH(MAX(CAST(TRX_DATETIME AS TIMESTAMP))) as last_trx_timestamp,
        COUNT(TrxID) as total_trx_count,
        SUM(TRX_AMT) as total_trx_amt,
        COUNT(CASE WHEN CAST(TRX_DATETIME AS TIMESTAMP) >= TIMESTAMP '2024-03-24' THEN TrxID END) as trx_cnt_7d,
        COUNT(CASE WHEN CAST(TRX_DATETIME AS TIMESTAMP) >= TIMESTAMP '2024-03-01' THEN TrxID END) as trx_cnt_30d,
        SUM(CASE WHEN CAST(TRX_DATETIME AS TIMESTAMP) >= TIMESTAMP '2024-03-24' THEN TRX_AMT END) as trx_amt_7d,
        SUM(CASE WHEN CAST(TRX_DATETIME AS TIMESTAMP) >= TIMESTAMP '2024-03-01' THEN TRX_AMT END) as trx_amt_30d,
        COUNT(CASE WHEN TRX_TYPE = 'Bill' THEN TrxID END) * 1.0 / NULLIF(COUNT(TrxID), 0) as bill_ratio,
        COUNT(CASE WHEN TRX_TYPE = 'Merchant' THEN TrxID END) * 1.0 / NULLIF(COUNT(TrxID), 0) as merchant_ratio,
        COUNT(DISTINCT DST_ACCOUNT) as unique_destinations
    FROM read_parquet('{trx_glob}')
    GROUP BY SRC_ACCOUNT
),
bal_features AS (
    SELECT
        ACCOUNT_ID,
        AVG(AVAILABLE_BALANCE) as avg_balance,
        STDDEV(AVAILABLE_BALANCE) as bal_volatility,
        MIN(AVAILABLE_BALANCE) as min_balance,
        MAX(AVAILABLE_BALANCE) as max_balance,
        AVG(CASE WHEN CAST(DATE AS DATE) >= CAST('2024-03-24' AS DATE) THEN AVAILABLE_BALANCE END) as avg_bal_7d
    FROM read_parquet('{bal_glob}')
    GROUP BY ACCOUNT_ID
)
SELECT
    k.ACCOUNT_ID,
    k.GENDER,
    k.REGION,
    EPOCH(CAST(k.ACCOUNT_OPEN_DATE AS TIMESTAMP)) as account_age_seconds,
    t.last_trx_timestamp,
    t.total_trx_count,
    t.total_trx_amt,
    t.trx_cnt_7d,
    t.trx_cnt_30d,
    t.trx_amt_7d,
    t.trx_amt_30d,
    t.bill_ratio,
    t.merchant_ratio,
    t.unique_destinations,
    b.avg_balance,
    b.bal_volatility,
    b.min_balance,
    b.max_balance,
    b.avg_bal_7d,
    (t.trx_cnt_7d * 1.0 / NULLIF(t.trx_cnt_30d, 0)) as velocity_cnt_ratio,
    (b.avg_bal_7d * 1.0 / NULLIF(b.avg_balance, 0)) as balance_decay_ratio
FROM base_kyc k
LEFT JOIN trx_features t ON k.ACCOUNT_ID = t.ACCOUNT_ID
LEFT JOIN bal_features b ON k.ACCOUNT_ID = b.ACCOUNT_ID
"""

# 4. EXECUTE AND STREAM INTO POLARS
df_features = con.execute(query).pl()

# Handle Nulls & Categoricals in Polars
df_features = df_features.with_columns([
    pl.col("GENDER").cast(pl.Categorical),
    pl.col("REGION").cast(pl.Categorical),
]).fill_null(0)

print(f"Features Extraction Complete successfully! Shape: {df_features.shape}")
```

```python
# ==========================================
# STAGE 3 & 4: SPLITTING & PREPARING DATA
# ==========================================
df_train_labels = pl.read_csv(train_labels_path)
df_test_labels = pl.read_csv(test_path)

# Merge back to target frameworks
train_pl = df_train_labels.join(df_features, on="ACCOUNT_ID", how="inner")
test_pl = df_test_labels.join(df_features, on="ACCOUNT_ID", how="left").fill_null(0)

# Convert to Pandas for modeling interface compatibility
train_df = train_pl.to_pandas()
test_df = test_pl.to_pandas()

features = [c for c in train_df.columns if c not in ["ACCOUNT_ID", "CHURN"]]
cat_features = ["GENDER", "REGION"]

for col in cat_features:
    train_df[col] = train_df[col].astype('category')
    test_df[col] = test_df[col].astype('category')

X = train_df[features]
y = train_df["CHURN"]
X_test = test_df[features]

print(f"Train Shape: {X.shape} | Churn Rate: {y.mean():.4f}")
```

```python
# ==========================================
# STAGE 5 & 6: STRATIFIED K-FOLD & BLENDING (LGBM + CATBOOST)
# ==========================================
n_splits = 5
skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)

lgb_oof = np.zeros(len(train_df))
cat_oof = np.zeros(len(train_df))
lgb_preds = np.zeros(len(test_df))
cat_preds = np.zeros(len(test_df))

lgb_params = {
    'objective': 'binary',
    'metric': 'auc',
    'learning_rate': 0.05,
    'num_leaves': 63,
    'max_depth': -1,
    'feature_fraction': 0.8,
    'bagging_fraction': 0.8,
    'bagging_freq': 1,
    'verbose': -1,
    'random_state': 42,
    'n_estimators': 3000
}

for fold, (train_idx, val_idx) in enumerate(skf.split(X, y)):
    print(f"--- Training Fold {fold + 1} ---")
    X_train, y_train = X.iloc[train_idx], y.iloc[train_idx]
    X_val, y_val = X.iloc[val_idx], y.iloc[val_idx]

    # 1. LightGBM
    model_lgb = lgb.LGBMClassifier(**lgb_params)
    model_lgb.fit(
        X_train, y_train,
        eval_set=[(X_val, y_val)],
        callbacks=[lgb.early_stopping(stopping_rounds=100, verbose=False)]
    )

    lgb_oof[val_idx] = model_lgb.predict_proba(X_val)[:, 1]
    lgb_preds += model_lgb.predict_proba(X_test)[:, 1] / n_splits

    # 2. CatBoost
    model_cat = CatBoostClassifier(
        iterations=2500,
        learning_rate=0.05,
        depth=6,
        eval_metric='AUC',
        random_seed=42,
        verbose=False,
        cat_features=cat_features
    )
    model_cat.fit(
        X_train, y_train,
        eval_set=(X_val, y_val),
        early_stopping_rounds=100
    )

    cat_oof[val_idx] = model_cat.predict_proba(X_val)[:, 1]
    cat_preds += model_cat.predict_proba(X_test)[:, 1] / n_splits

# Evaluate Out-Of-Fold (OOF) Scores
lgb_cv = roc_auc_score(y, lgb_oof)
cat_cv = roc_auc_score(y, cat_oof)
blend_oof = (lgb_oof * 0.5) + (cat_oof * 0.5)
blend_cv = roc_auc_score(y, blend_oof)

print("\n==========================================")
print(f"LightGBM CV ROC-AUC: {lgb_cv:.5f}")
print(f"CatBoost CV ROC-AUC: {cat_cv:.5f}")
print(f"Rank Averaged Blend CV ROC-AUC: {blend_cv:.5f}")
print("==========================================")
```

```python
# ==========================================
# STAGE 8: GENERATING FINAL SUBMISSION
# ==========================================
final_preds = (lgb_preds * 0.5) + (cat_preds * 0.5)

submission = pl.DataFrame({
    "ACCOUNT_ID": test_df["ACCOUNT_ID"],
    "CHURN_PROB": final_preds
})

submission.write_csv("predictions.csv")
print("Submission successfully saved to 'predictions.csv'!")
```

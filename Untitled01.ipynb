```python
trx_feats = con.execute(f"""
WITH t AS (
  SELECT SRC_ACCOUNT AS ACCOUNT_ID, DST_ACCOUNT, TRX_TYPE, TRX_AMT,
         CAST(TRX_DATETIME AS TIMESTAMP) AS TS
  FROM read_parquet('{TRX}')
)
SELECT ACCOUNT_ID,
  count(*)                          AS trx_cnt_total,
  sum(TRX_AMT)                      AS trx_amt_total,
  avg(TRX_AMT)                      AS trx_amt_mean,
  stddev(TRX_AMT)                   AS trx_amt_std,
  max(TRX_AMT)                      AS trx_amt_max,
  median(TRX_AMT)                   AS trx_amt_med,
  count(*) FILTER (TS >= TIMESTAMP '{ref_str}' - INTERVAL 7  DAY) AS trx_cnt_7d,
  count(*) FILTER (TS >= TIMESTAMP '{ref_str}' - INTERVAL 14 DAY) AS trx_cnt_14d,
  count(*) FILTER (TS >= TIMESTAMP '{ref_str}' - INTERVAL 30 DAY) AS trx_cnt_30d,
  count(*) FILTER (TS >= TIMESTAMP '{ref_str}' - INTERVAL 60 DAY) AS trx_cnt_60d,
  sum(TRX_AMT) FILTER (TS >= TIMESTAMP '{ref_str}' - INTERVAL 30 DAY) AS trx_amt_30d,
  count(*) FILTER (TRX_TYPE='P2P')      * 1.0 / count(*) AS p2p_ratio,
  count(*) FILTER (TRX_TYPE='Merchant') * 1.0 / count(*) AS merchant_ratio,
  count(*) FILTER (TRX_TYPE='Bill')     * 1.0 / count(*) AS bill_ratio,
  date_diff('day', max(TS), TIMESTAMP '{ref_str}') AS recency_days,
  date_diff('day', min(TS), TIMESTAMP '{ref_str}') AS first_seen_days,
  count(DISTINCT DST_ACCOUNT)      AS n_counterparties,
  count(DISTINCT date_trunc('day', TS)) AS active_days
FROM t GROUP BY 1
""").df()
print("trx_feats:", trx_feats.shape)
```

```python
trx_in = con.execute(f"""
SELECT DST_ACCOUNT AS ACCOUNT_ID,
  count(*) AS trx_in_cnt, sum(TRX_AMT) AS trx_in_amt,
  count(DISTINCT SRC_ACCOUNT) AS n_senders,
  date_diff('day', max(CAST(TRX_DATETIME AS TIMESTAMP)), TIMESTAMP '{ref_str}') AS in_recency_days
FROM read_parquet('{TRX}') GROUP BY 1
""").df()
print("trx_in:", trx_in.shape)

last_activity = con.execute(f"""
WITH t AS (
  SELECT SRC_ACCOUNT AS ACCOUNT_ID, CAST(TRX_DATETIME AS TIMESTAMP) AS TS, TRX_AMT
  FROM read_parquet('{TRX}')
), ranked AS (
  SELECT ACCOUNT_ID, TS, TRX_AMT,
         ROW_NUMBER() OVER (PARTITION BY ACCOUNT_ID ORDER BY TS DESC) AS rn
  FROM t
)
SELECT ACCOUNT_ID,
  avg(TRX_AMT) FILTER (rn <= 3)  AS last3_amt_avg,
  avg(TRX_AMT) FILTER (rn <= 3) / (avg(TRX_AMT) FILTER (rn > 3 AND rn <= 13) + 1) AS amt_drop_ratio,
  date_diff('day', max(TS) FILTER (rn = 2), max(TS) FILTER (rn = 1)) AS last_gap_days,
  date_diff('day', min(TS), max(TS)) * 1.0 / (count(*) + 1) AS avg_gap_days
FROM ranked GROUP BY 1
""").df()
print("last_activity:", last_activity.shape)

# monthly trend over the 3 available months (m1 = most recent 30d)
monthly = con.execute(f"""
WITH t AS (
  SELECT SRC_ACCOUNT AS ACCOUNT_ID, TRX_AMT, CAST(TRX_DATETIME AS TIMESTAMP) AS TS
  FROM read_parquet('{TRX}')
)
SELECT ACCOUNT_ID,
  count(*) FILTER (TS >= TIMESTAMP '{ref_str}' - INTERVAL 30 DAY) AS cnt_m1,
  count(*) FILTER (TS BETWEEN TIMESTAMP '{ref_str}' - INTERVAL 60 DAY
                          AND TIMESTAMP '{ref_str}' - INTERVAL 30 DAY) AS cnt_m2,
  count(*) FILTER (TS <  TIMESTAMP '{ref_str}' - INTERVAL 60 DAY) AS cnt_m3,
  (count(*) FILTER (TS >= TIMESTAMP '{ref_str}' - INTERVAL 30 DAY)
   - count(*) FILTER (TS < TIMESTAMP '{ref_str}' - INTERVAL 60 DAY)) * 1.0 / 2 AS cnt_slope,
  CASE WHEN count(*) FILTER (TS >= TIMESTAMP '{ref_str}' - INTERVAL 30 DAY) > 0
        AND count(*) FILTER (TS BETWEEN TIMESTAMP '{ref_str}' - INTERVAL 60 DAY
                                    AND TIMESTAMP '{ref_str}' - INTERVAL 30 DAY) > 0
        AND count(*) FILTER (TS < TIMESTAMP '{ref_str}' - INTERVAL 60 DAY) > 0
       THEN 1 ELSE 0 END AS active_all_3_months
FROM t GROUP BY 1
""").df()
print("monthly:", monthly.shape)
```

```python
bal_feats = con.execute(f"""
WITH b AS (
  SELECT ACCOUNT_ID, AVAILABLE_BALANCE, CAST(DATE AS DATE) AS D,
         ROW_NUMBER() OVER (PARTITION BY ACCOUNT_ID ORDER BY CAST(DATE AS DATE) DESC) AS rn
  FROM read_parquet('{BAL}')
)
SELECT ACCOUNT_ID,
  avg(AVAILABLE_BALANCE)    AS bal_mean,
  stddev(AVAILABLE_BALANCE) AS bal_std,
  min(AVAILABLE_BALANCE)    AS bal_min,
  max(AVAILABLE_BALANCE)    AS bal_max,
  arg_max(AVAILABLE_BALANCE, D) AS bal_last,
  arg_min(AVAILABLE_BALANCE, D) AS bal_first,
  count(*) FILTER (AVAILABLE_BALANCE <= 0) * 1.0 / count(*) AS zero_bal_ratio,
  avg(AVAILABLE_BALANCE) FILTER (rn <= 7)  AS bal_last7,
  avg(AVAILABLE_BALANCE) FILTER (rn <= 30) AS bal_last30,
  avg(AVAILABLE_BALANCE) FILTER (rn > 30 AND rn <= 60) AS bal_prev30,
  avg(AVAILABLE_BALANCE) FILTER (rn <= 30)
    - avg(AVAILABLE_BALANCE) FILTER (rn > 30 AND rn <= 60) AS bal_mom_change,
  count(*) FILTER (rn <= 30 AND AVAILABLE_BALANCE <= 0) AS zero_days_last30,
  stddev(AVAILABLE_BALANCE) FILTER (rn <= 30) AS bal_std_last30
FROM b GROUP BY 1
""").df()
print("bal_feats:", bal_feats.shape)

# network diversity
network = con.execute(f"""
SELECT SRC_ACCOUNT AS ACCOUNT_ID,
  count(DISTINCT DST_ACCOUNT) AS unique_receivers,
  count(DISTINCT DST_ACCOUNT) * 1.0 / (count(*) + 1) AS diversity_ratio,
  count(DISTINCT TRX_TYPE) AS trx_type_diversity
FROM read_parquet('{TRX}') GROUP BY 1
""").df()
print("network:", network.shape)

kyc = con.execute(f"""
SELECT ACCOUNT_ID, GENDER, REGION,
  date_diff('day', CAST(ACCOUNT_OPEN_DATE AS DATE), DATE '{ref_str}') AS account_age_days
FROM read_parquet('{KYC}') WHERE ACCOUNT_TYPE = 'Customer'
""").df()
print("kyc:", kyc.shape)
```

```python
train_labels = pd.read_csv(f"{DATA_DIR}/train_labels.csv")
test = pd.read_csv(f"{DATA_DIR}/test.csv")
label_col = "CHURN"

def build(df):
    df = (df.merge(kyc, on="ACCOUNT_ID", how="left")
            .merge(trx_feats, on="ACCOUNT_ID", how="left")
            .merge(trx_in, on="ACCOUNT_ID", how="left")
            .merge(last_activity, on="ACCOUNT_ID", how="left")
            .merge(monthly, on="ACCOUNT_ID", how="left")
            .merge(bal_feats, on="ACCOUNT_ID", how="left")
            .merge(network, on="ACCOUNT_ID", how="left"))

    # log transforms + zero flags for skewed counts/amounts
    for c in ["trx_cnt_total","trx_amt_total","trx_in_cnt","trx_in_amt",
              "unique_receivers","cnt_m1","cnt_m2","cnt_m3","n_counterparties"]:
        if c in df.columns:
            df[f"{c}_is_zero"] = df[c].isna().astype(int)
            df[c] = np.log1p(df[c].fillna(0))

    # recency-type fills (no activity = large value)
    for c in ["recency_days","in_recency_days","last_gap_days","avg_gap_days","first_seen_days"]:
        if c in df.columns: df[c] = df[c].fillna(999)

    # ratio / trend features
    df["trend_7v30"]   = df["trx_cnt_7d"].fillna(0) / (df["trx_cnt_30d"].fillna(0) + 1)
    df["trend_30v60"]  = df["trx_cnt_30d"].fillna(0) / (df["trx_cnt_60d"].fillna(0) + 1)
    df["m1_vs_m2"]     = df["cnt_m1"].fillna(0) / (df["cnt_m2"].fillna(0) + 1)
    df["bal_drop"]     = (df["bal_mom_change"].fillna(0) < 0).astype(int)
    df["bal_last_vs_mean"] = df["bal_last"].fillna(0) / (df["bal_mean"].fillna(0).abs() + 1)
    df["activity_density"] = df["active_days"].fillna(0) / (df["first_seen_days"].replace(0,1) + 1)

    df = df.fillna(0)
    for c in ["GENDER","REGION"]:
        if c in df.columns: df[c] = df[c].astype("category")
    return df

train_df = build(train_labels.copy())
test_df  = build(test[["ACCOUNT_ID"]].copy())
features = [c for c in train_df.columns if c not in ["ACCOUNT_ID", label_col]]
print("features:", len(features), "| churn rate:", train_df[label_col].mean())
```

```python
import lightgbm as lgb, xgboost as xgb
from catboost import CatBoostClassifier
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_auc_score
from scipy.stats import rankdata

X, y = train_df[features], train_df[label_col].values
X_test = test_df[features]
cat_feats = [c for c in ["GENDER","REGION"] if c in features]

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
oof = {m: np.zeros(len(X)) for m in ["lgb","xgb","cat"]}
tst = {m: np.zeros(len(X_test)) for m in ["lgb","xgb","cat"]}

for fold,(tr,va) in enumerate(skf.split(X,y)):
    Xtr,Xva,ytr,yva = X.iloc[tr],X.iloc[va],y[tr],y[va]
    pw = (ytr==0).sum()/(ytr==1).sum()

    m = lgb.LGBMClassifier(n_estimators=6000, learning_rate=0.01, num_leaves=127,
        subsample=0.8, colsample_bytree=0.7, min_child_samples=30,
        reg_alpha=0.1, reg_lambda=1.0, scale_pos_weight=pw,
        random_state=42+fold, n_jobs=-1)
    m.fit(Xtr,ytr, eval_set=[(Xva,yva)], eval_metric="auc",
          callbacks=[lgb.early_stopping(200, verbose=False)],
          categorical_feature=cat_feats)
    oof["lgb"][va]=m.predict_proba(Xva)[:,1]; tst["lgb"]+=m.predict_proba(X_test)[:,1]/5

    Xtr_x,Xva_x,Xte_x = Xtr.copy(),Xva.copy(),X_test.copy()
    for c in cat_feats:
        Xtr_x[c]=Xtr_x[c].cat.codes; Xva_x[c]=Xva_x[c].cat.codes; Xte_x[c]=Xte_x[c].cat.codes
    xm = xgb.XGBClassifier(n_estimators=6000, learning_rate=0.01, max_depth=7,
        subsample=0.8, colsample_bytree=0.7, scale_pos_weight=pw,
        eval_metric="auc", early_stopping_rounds=200,
        random_state=42+fold, tree_method="hist", verbosity=0)
    xm.fit(Xtr_x,ytr, eval_set=[(Xva_x,yva)], verbose=False)
    oof["xgb"][va]=xm.predict_proba(Xva_x)[:,1]; tst["xgb"]+=xm.predict_proba(Xte_x)[:,1]/5

    cm = CatBoostClassifier(iterations=6000, learning_rate=0.02, depth=7,
        l2_leaf_reg=3.0, scale_pos_weight=pw, eval_metric="AUC",
        random_seed=42+fold, verbose=False, cat_features=cat_feats)
    cm.fit(Xtr,ytr, eval_set=(Xva,yva), early_stopping_rounds=200, use_best_model=True)
    oof["cat"][va]=cm.predict_proba(Xva)[:,1]; tst["cat"]+=cm.predict_proba(X_test)[:,1]/5

    print(f"Fold{fold+1} LGB {roc_auc_score(yva,oof['lgb'][va]):.5f} "
          f"XGB {roc_auc_score(yva,oof['xgb'][va]):.5f} "
          f"CAT {roc_auc_score(yva,oof['cat'][va]):.5f}")

for k in oof: print(f"OOF {k}: {roc_auc_score(y,oof[k]):.5f}")
```

```python
# rank-normalize OOF and test, search best 3-way weights on OOF
ro = {k: rankdata(oof[k])/len(oof[k]) for k in oof}
rt = {k: rankdata(tst[k])/len(tst[k]) for k in tst}

best=(0,None)
for a in np.arange(0,1.01,0.1):
    for b in np.arange(0,1.01-a,0.1):
        c=1-a-b
        s=roc_auc_score(y, a*ro["lgb"]+b*ro["xgb"]+c*ro["cat"])
        if s>best[0]: best=(s,(round(a,2),round(b,2),round(c,2)))
print("Best OOF AUC:", best[0], "weights(lgb,xgb,cat):", best[1])

a,b,c = best[1]
final = a*rt["lgb"]+b*rt["xgb"]+c*rt["cat"]

sub = pd.DataFrame({"ACCOUNT_ID": test_df["ACCOUNT_ID"], "CHURN_PROB": final})
sub.to_csv("/content/predictions.csv", index=False)
print(sub.head(), "\nrows:", len(sub))

from google.colab import files
files.download("/content/predictions.csv")
```

```python
# ===== STAGE 2A-1: light aggregations (no DISTINCT) =====
trx_out1 = con.execute(f"""
WITH t AS (
  SELECT SRC_ACCOUNT AS ACCOUNT_ID, TRX_TYPE, TRX_AMT,
         CAST(TRX_DATETIME AS TIMESTAMP) AS TS,
         substr(DST_ACCOUNT,1,4) AS DST_PFX
  FROM read_parquet('{TRX}')
)
SELECT ACCOUNT_ID,
  count(*)        AS trx_cnt_total,
  sum(TRX_AMT)    AS trx_amt_total,
  avg(TRX_AMT)    AS trx_amt_mean,
  stddev(TRX_AMT) AS trx_amt_std,
  max(TRX_AMT)    AS trx_amt_max,
  median(TRX_AMT) AS trx_amt_med,
  count(*) FILTER (TS >= TIMESTAMP '{REF}' - INTERVAL 7  DAY) AS cnt_7d,
  count(*) FILTER (TS >= TIMESTAMP '{REF}' - INTERVAL 14 DAY) AS cnt_14d,
  count(*) FILTER (TS >= TIMESTAMP '{REF}' - INTERVAL 30 DAY) AS cnt_30d,
  count(*) FILTER (TS >= TIMESTAMP '{REF}' - INTERVAL 60 DAY) AS cnt_60d,
  count(*) FILTER (TS <  TIMESTAMP '{REF}' - INTERVAL 60 DAY) AS cnt_first30,
  sum(TRX_AMT) FILTER (TS >= TIMESTAMP '{REF}' - INTERVAL 30 DAY) AS amt_30d,
  sum(TRX_AMT) FILTER (TS >= TIMESTAMP '{REF}' - INTERVAL 7  DAY) AS amt_7d,
  count(*) FILTER (TRX_TYPE='P2P')         * 1.0 / count(*) AS r_p2p,
  count(*) FILTER (TRX_TYPE='MerchantPay') * 1.0 / count(*) AS r_merchant,
  count(*) FILTER (TRX_TYPE='BillPay')     * 1.0 / count(*) AS r_bill,
  count(*) FILTER (TRX_TYPE='CashIn')      * 1.0 / count(*) AS r_cashin,
  count(*) FILTER (TRX_TYPE='CashOut')     * 1.0 / count(*) AS r_cashout,
  count(*) FILTER (DST_PFX='MRCH') * 1.0 / count(*) AS dst_merch_ratio,
  count(*) FILTER (DST_PFX='BILL') * 1.0 / count(*) AS dst_bill_ratio,
  date_diff('day', max(TS), TIMESTAMP '{REF}') AS recency_days,
  date_diff('day', min(TS), TIMESTAMP '{REF}') AS first_seen_days,
  avg(extract('hour' FROM TS))    AS hour_mean,
  stddev(extract('hour' FROM TS)) AS hour_std,
  count(*) FILTER (extract('dow' FROM TS) IN (0,6)) * 1.0 / count(*) AS weekend_ratio
FROM t GROUP BY 1
""").df()
print("trx_out1:", trx_out1.shape)
display(trx_out1.head(3))
```

```python
# ===== STAGE 2A-2: DISTINCT-based features (heavier, run alone) =====
trx_out2 = con.execute(f"""
WITH t AS (
  SELECT SRC_ACCOUNT AS ACCOUNT_ID, DST_ACCOUNT, TRX_TYPE,
         CAST(TRX_DATETIME AS TIMESTAMP) AS TS
  FROM read_parquet('{TRX}')
)
SELECT ACCOUNT_ID,
  count(DISTINCT CASE WHEN DST_ACCOUNT<>ACCOUNT_ID THEN DST_ACCOUNT END) AS n_counterparties,
  count(DISTINCT date_trunc('day', TS)) AS active_days,
  count(DISTINCT TRX_TYPE)              AS type_diversity
FROM t GROUP BY 1
""").df()
print("trx_out2:", trx_out2.shape)
display(trx_out2.head(3))
```

```python
# ===== STAGE 2B-1: gap & streak features (window functions) =====
gap_feats = con.execute(f"""
WITH t AS (
  SELECT SRC_ACCOUNT AS ACCOUNT_ID, TRX_AMT,
         CAST(TRX_DATETIME AS TIMESTAMP) AS TS
  FROM read_parquet('{TRX}')
),
g AS (
  SELECT ACCOUNT_ID, TS, TRX_AMT,
    date_diff('day', lag(TS) OVER (PARTITION BY ACCOUNT_ID ORDER BY TS), TS) AS gap_days,
    ROW_NUMBER() OVER (PARTITION BY ACCOUNT_ID ORDER BY TS DESC) AS rn_desc
  FROM t
)
SELECT ACCOUNT_ID,
  avg(gap_days)                          AS gap_mean,
  stddev(gap_days)                       AS gap_std,
  max(gap_days)                          AS gap_max,          -- longest inactivity streak
  date_diff('day', max(TS) FILTER (rn_desc=2), max(TS) FILTER (rn_desc=1)) AS last_gap_days,
  avg(TRX_AMT) FILTER (rn_desc <= 3)     AS last3_amt_avg,
  avg(TRX_AMT) FILTER (rn_desc <= 3)
    / (avg(TRX_AMT) FILTER (rn_desc > 3 AND rn_desc <= 13) + 1) AS amt_drop_ratio
FROM g GROUP BY 1
""").df()
print("gap_feats:", gap_feats.shape)
display(gap_feats.head(3))
```

```python
# ===== STAGE 2B-2: INCOMING features (account as DST/receiver) =====
trx_in = con.execute(f"""
WITH t AS (
  SELECT DST_ACCOUNT AS ACCOUNT_ID, SRC_ACCOUNT, TRX_AMT,
         CAST(TRX_DATETIME AS TIMESTAMP) AS TS
  FROM read_parquet('{TRX}')
  WHERE SRC_ACCOUNT <> DST_ACCOUNT      -- exclude self-txn
)
SELECT ACCOUNT_ID,
  count(*)                 AS in_cnt,
  sum(TRX_AMT)             AS in_amt,
  avg(TRX_AMT)             AS in_amt_mean,
  count(*) FILTER (TS >= TIMESTAMP '{REF}' - INTERVAL 30 DAY) AS in_cnt_30d,
  count(*) FILTER (TS >= TIMESTAMP '{REF}' - INTERVAL 7  DAY) AS in_cnt_7d,
  date_diff('day', max(TS), TIMESTAMP '{REF}') AS in_recency_days
FROM t GROUP BY 1
""").df()
print("trx_in:", trx_in.shape)
display(trx_in.head(3))
```

```python
# ===== STAGE 2C: balance features =====
bal_feats = con.execute(f"""
WITH b AS (
  SELECT ACCOUNT_ID, AVAILABLE_BALANCE, CAST(DATE AS DATE) AS D,
         ROW_NUMBER() OVER (PARTITION BY ACCOUNT_ID ORDER BY CAST(DATE AS DATE) DESC) AS rn
  FROM read_parquet('{BAL}')
)
SELECT ACCOUNT_ID,
  avg(AVAILABLE_BALANCE)    AS bal_mean,
  stddev(AVAILABLE_BALANCE) AS bal_std,
  min(AVAILABLE_BALANCE)    AS bal_min,
  max(AVAILABLE_BALANCE)    AS bal_max,
  arg_max(AVAILABLE_BALANCE, D) AS bal_last,
  arg_min(AVAILABLE_BALANCE, D) AS bal_at_first,
  count(*) FILTER (AVAILABLE_BALANCE <= 0) * 1.0 / count(*) AS zero_bal_ratio,
  avg(AVAILABLE_BALANCE) FILTER (rn <= 7)               AS bal_last7,
  avg(AVAILABLE_BALANCE) FILTER (rn <= 30)              AS bal_last30,
  avg(AVAILABLE_BALANCE) FILTER (rn > 30 AND rn <= 60)  AS bal_prev30,
  avg(AVAILABLE_BALANCE) FILTER (rn <= 30)
    - avg(AVAILABLE_BALANCE) FILTER (rn > 30 AND rn <= 60) AS bal_mom_change,
  count(*) FILTER (rn <= 30 AND AVAILABLE_BALANCE <= 0)  AS zero_days_last30,
  stddev(AVAILABLE_BALANCE) FILTER (rn <= 30)           AS bal_std_last30,
  -- linear trend slope over the full window (corr * std ratio); negative = declining
  regr_slope(AVAILABLE_BALANCE, date_diff('day', DATE '2024-01-01', D)) AS bal_slope,
  count(*)                                              AS bal_days
FROM b GROUP BY 1
""").df()
print("bal_feats:", bal_feats.shape)
display(bal_feats.head(3))
```

```python
# ===== STAGE 2D: KYC features =====
kyc = con.execute(f"""
SELECT ACCOUNT_ID, GENDER, REGION,
  date_diff('day', CAST(ACCOUNT_OPEN_DATE AS DATE), DATE '{REF}') AS account_age_days
FROM read_parquet('{KYC}')
WHERE ACCOUNT_TYPE = 'Customer'
""").df()
print("kyc:", kyc.shape)
display(kyc.head(3))
```

```python
# ===== STAGE 3: merge all tables + build engineered features =====
train_labels = pd.read_csv(f"{DATA_DIR}/train_labels.csv")
test = pd.read_csv(f"{DATA_DIR}/test.csv")
label_col = "CHURN"

def build(df):
    df = (df.merge(kyc,       on="ACCOUNT_ID", how="left")
            .merge(trx_out1,  on="ACCOUNT_ID", how="left")
            .merge(trx_out2,  on="ACCOUNT_ID", how="left")
            .merge(gap_feats, on="ACCOUNT_ID", how="left")
            .merge(trx_in,    on="ACCOUNT_ID", how="left")
            .merge(bal_feats, on="ACCOUNT_ID", how="left"))

    # ---- "no activity at all" flag (strong churn signal) ----
    df["no_trx"] = df["trx_cnt_total"].isna().astype(int)
    df["no_incoming"] = df["in_cnt"].isna().astype(int)

    # ---- log transforms + zero flags for skewed counts/amounts ----
    for c in ["trx_cnt_total","trx_amt_total","trx_amt_max","amt_30d","amt_7d",
              "in_cnt","in_amt","n_counterparties","cnt_30d","cnt_60d","cnt_first30"]:
        if c in df.columns:
            df[c] = np.log1p(df[c].fillna(0).clip(lower=0))

    # ---- recency-type fills (no activity = large number) ----
    for c in ["recency_days","in_recency_days","last_gap_days","gap_max","gap_mean",
              "first_seen_days"]:
        if c in df.columns: df[c] = df[c].fillna(999)

    # ---- trend / ratio features (churn = declining activity) ----
    df["trend_7v30"]   = df["cnt_7d"].fillna(0)  / (df["cnt_30d_raw"].fillna(0) + 1) if "cnt_30d_raw" in df else \
                         df["cnt_7d"].fillna(0)  / (np.expm1(df["cnt_30d"].fillna(0)) + 1)
    df["trend_30v60"]  = np.expm1(df["cnt_30d"].fillna(0)) / (np.expm1(df["cnt_60d"].fillna(0)) + 1)
    df["recent_vs_first"] = np.expm1(df["cnt_30d"].fillna(0)) / (np.expm1(df["cnt_first30"].fillna(0)) + 1)

    # ---- balance interactions ----
    df["bal_drop_flag"]    = (df["bal_mom_change"].fillna(0) < 0).astype(int)
    df["bal_last_vs_mean"] = df["bal_last"].fillna(0) / (df["bal_mean"].fillna(0).abs() + 1)
    df["bal_slope_neg"]    = (df["bal_slope"].fillna(0) < 0).astype(int)

    # ---- activity density ----
    df["activity_density"] = df["active_days"].fillna(0) / (df["first_seen_days"].replace(0,1) + 1)
    df["amt_per_trx_recent"] = df["amt_30d"].fillna(0) / (np.expm1(df["cnt_30d"].fillna(0)) + 1)

    # ---- final fill + categoricals ----
    df = df.fillna(0)
    for c in ["GENDER","REGION"]:
        if c in df.columns: df[c] = df[c].astype("category")
    return df

train_df = build(train_labels.copy())
test_df  = build(test[["ACCOUNT_ID"]].copy())

features = [c for c in train_df.columns if c not in ["ACCOUNT_ID", label_col]]
print("n features:", len(features))
print("churn rate:", train_df[label_col].mean())
print("train shape:", train_df.shape, "| test shape:", test_df.shape)
print("any NaN train:", train_df[features].isna().sum().sum(), "| test:", test_df[features].isna().sum().sum())
```

```python
# ===== STAGE 4: LightGBM 5-fold CV =====
import lightgbm as lgb
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_auc_score, average_precision_score

X, y = train_df[features], train_df[label_col].values
X_test = test_df[features]
cat_feats = [c for c in ["GENDER","REGION"] if c in features]

params = dict(
    n_estimators=10000,
    learning_rate=0.02,
    num_leaves=63,
    max_depth=-1,
    min_child_samples=100,
    subsample=0.8,
    subsample_freq=1,
    colsample_bytree=0.8,
    reg_alpha=0.5,
    reg_lambda=2.0,
    objective="binary",
    metric="auc",
    random_state=42,
    n_jobs=-1,
    verbose=-1,
)

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
oof = np.zeros(len(X))
pred = np.zeros(len(X_test))
best_iters = []

for fold,(tr,va) in enumerate(skf.split(X,y)):
    Xtr,Xva,ytr,yva = X.iloc[tr],X.iloc[va],y[tr],y[va]
    m = lgb.LGBMClassifier(**params)
    m.fit(Xtr,ytr, eval_set=[(Xva,yva)], eval_metric="auc",
          callbacks=[lgb.early_stopping(300, verbose=False)],
          categorical_feature=cat_feats)
    oof[va] = m.predict_proba(Xva)[:,1]
    pred   += m.predict_proba(X_test)[:,1] / 5
    best_iters.append(m.best_iteration_)
    print(f"Fold{fold+1}  AUC {roc_auc_score(yva,oof[va]):.5f}  best_iter {m.best_iteration_}")

print("\n==============================")
print(f"OOF AUC:  {roc_auc_score(y,oof):.5f}")
print(f"OOF AvgP: {average_precision_score(y,oof):.5f}")
print("avg best_iter:", int(np.mean(best_iters)))
print("==============================")
```

```python
# ===== STAGE 4B: diagnose - what drives churn? =====
import lightgbm as lgb

params_diag = {**params, "n_estimators": 300}
m = lgb.LGBMClassifier(**params_diag)
m.fit(X, y, categorical_feature=cat_feats)

imp = pd.DataFrame({"feature": features, "gain": m.booster_.feature_importance("gain")})
imp = imp.sort_values("gain", ascending=False).head(20)
print("TOP 20 FEATURES BY GAIN:")
print(imp.to_string(index=False))

# churn rate vs recency  (is churn = inactivity?)
print("\n--- churn rate by recency_days bucket ---")
tmp = train_df[["recency_days", label_col]].copy()
tmp["bucket"] = pd.cut(tmp["recency_days"], [-1,0,1,3,7,14,30,60,1000])
print(tmp.groupby("bucket", observed=True)[label_col].agg(["mean","count"]))

# churn rate vs account_age
print("\n--- churn rate by account_age_days bucket ---")
tmp2 = train_df[["account_age_days", label_col]].copy()
tmp2["bucket"] = pd.qcut(tmp2["account_age_days"], 8, duplicates="drop")
print(tmp2.groupby("bucket", observed=True)[label_col].agg(["mean","count"]))

# churn vs no_trx / region
print("\n--- churn by no_trx ---")
print(train_df.groupby("no_trx")[label_col].agg(["mean","count"]))
print("\n--- churn by REGION ---")
print(train_df.groupby("REGION", observed=True)[label_col].agg(["mean","count"]).sort_values("mean"))
```

```python
# ===== STAGE 4C: balance-change recency (independent activity signal) =====
bal_activity = con.execute(f"""
WITH b AS (
  SELECT ACCOUNT_ID, AVAILABLE_BALANCE, CAST(DATE AS DATE) AS D,
         AVAILABLE_BALANCE - lag(AVAILABLE_BALANCE) OVER (PARTITION BY ACCOUNT_ID ORDER BY CAST(DATE AS DATE)) AS dchg
  FROM read_parquet('{BAL}')
)
SELECT ACCOUNT_ID,
  date_diff('day', max(D) FILTER (abs(dchg) > 0.001), DATE '{REF}') AS bal_last_change_recency,
  count(*) FILTER (abs(dchg) > 0.001)                    AS bal_change_days,
  count(*) FILTER (abs(dchg) > 0.001 AND D >= DATE '{REF}' - INTERVAL 30 DAY) AS bal_change_30d,
  count(*) FILTER (abs(dchg) > 0.001 AND D >= DATE '{REF}' - INTERVAL 7  DAY) AS bal_change_7d,
  count(*) FILTER (abs(dchg) > 0.001 AND D >= DATE '{REF}' - INTERVAL 14 DAY) AS bal_change_14d
FROM b GROUP BY 1
""").df()
print("bal_activity:", bal_activity.shape)
display(bal_activity.head(3))

# does bal_change_recency add signal beyond trx recency?
chk = train_df[["ACCOUNT_ID","recency_days",label_col]].merge(
    bal_activity[["ACCOUNT_ID","bal_last_change_recency","bal_change_30d"]], on="ACCOUNT_ID", how="left")
chk["bal_last_change_recency"] = chk["bal_last_change_recency"].fillna(999)
print("\ncorr bal_change_recency vs churn:", chk["bal_last_change_recency"].corr(chk[label_col]))
print("corr trx recency vs churn:", chk["recency_days"].corr(chk[label_col]))

# key test: among recency 7-30 (noisy zone), does bal activity separate churners?
mask = (chk["recency_days"]>=7)&(chk["recency_days"]<=30)
sub_zone = chk[mask].copy()
sub_zone["bal_active_30d"] = (sub_zone["bal_change_30d"].fillna(0)>0).astype(int)
print("\n--- in noisy zone (recency 7-30): churn by balance activity ---")
print(sub_zone.groupby("bal_active_30d")[label_col].agg(["mean","count"]))
```

```python
# ===== STAGE 4D: what separates churners in the NOISY zone (recency 7-30)? =====
zone = train_df[(train_df["recency_days"]>=7)&(train_df["recency_days"]<=30)].copy()
print("noisy zone size:", len(zone), "churn rate:", zone[label_col].mean())

# train a model ONLY on this zone, see top features + achievable AUC
from sklearn.model_selection import cross_val_predict
import lightgbm as lgb
from sklearn.metrics import roc_auc_score

Xz = zone[features]; yz = zone[label_col].values
mz = lgb.LGBMClassifier(**{**params, "n_estimators": 400})
from sklearn.model_selection import StratifiedKFold
skf = StratifiedKFold(5, shuffle=True, random_state=42)
oofz = np.zeros(len(Xz))
for tr,va in skf.split(Xz,yz):
    mm = lgb.LGBMClassifier(**{**params,"n_estimators":400})
    mm.fit(Xz.iloc[tr],yz[tr],categorical_feature=cat_feats)
    oofz[va]=mm.predict_proba(Xz.iloc[va])[:,1]
print("zone-only OOF AUC:", roc_auc_score(yz,oofz))

mz.fit(Xz,yz,categorical_feature=cat_feats)
impz = pd.DataFrame({"feature":features,"gain":mz.booster_.feature_importance("gain")}).sort_values("gain",ascending=False).head(15)
print("\nTOP features in noisy zone:")
print(impz.to_string(index=False))

# within zone, churn rate by finer recency + recent count
zone["rec_fine"] = pd.cut(zone["recency_days"],[6,10,14,20,30])
print("\n--- churn by fine recency in zone ---")
print(zone.groupby("rec_fine",observed=True)[label_col].agg(["mean","count"]))

# does trend (declining) matter in zone?
zone["declining"] = (zone["trend_30v60"]<0.5).astype(int)
print("\n--- churn by declining(trend_30v60<0.5) in zone ---")
print(zone.groupby("declining")[label_col].agg(["mean","count"]))
```

```python
# ===== STAGE 4E: how much AUC from recency ALONE? + check for clean rules =====
from sklearn.metrics import roc_auc_score

# 1) recency alone (negative because high recency -> churn)
print("recency_days alone AUC:", roc_auc_score(y, train_df["recency_days"].values))
print("cnt_14d alone AUC:", roc_auc_score(y, -train_df["cnt_14d"].values))
print("cnt_30d alone AUC:", roc_auc_score(y, -train_df["cnt_30d"].values))

# 2) simple blend of the two strongest
blend = train_df["recency_days"].rank() - train_df["cnt_14d"].rank()
print("recency - cnt_14d blend AUC:", roc_auc_score(y, blend))

# 3) is there a perfectly clean region? churn=0 and churn=1 deterministic zones
print("\n--- recency_days==0 ---")
print(train_df[train_df["recency_days"]<=0][label_col].agg(["mean","count"]))
print("--- no_trx==1 ---")
print(train_df[train_df["no_trx"]==1][label_col].agg(["mean","count"]))

# 4) check last_gap_days sanity (possible bug)
print("\nlast_gap_days describe:")
print(train_df["last_gap_days"].describe())

# 5) MOST IMPORTANT: train full model but with ONLY clean features, compare
clean_feats = ["recency_days","cnt_7d","cnt_14d","cnt_30d","cnt_60d","cnt_first30",
               "trx_cnt_total","recent_vs_first","trend_30v60","trend_7v30",
               "active_days","gap_mean","gap_max","last_gap_days","first_seen_days",
               "no_trx","bal_std_last30","zero_days_last30"]
clean_feats = [c for c in clean_feats if c in features]
import lightgbm as lgb
from sklearn.model_selection import StratifiedKFold
skf = StratifiedKFold(5, shuffle=True, random_state=42)
oofc = np.zeros(len(train_df))
Xc = train_df[clean_feats]
for tr,va in skf.split(Xc,y):
    mm = lgb.LGBMClassifier(**{**params,"n_estimators":600})
    mm.fit(Xc.iloc[tr],y[tr])
    oofc[va]=mm.predict_proba(Xc.iloc[va])[:,1]
print("\nCLEAN-features-only OOF AUC:", roc_auc_score(y,oofc))
```

```python
# ===== STAGE 5A: fine-grained weekly activity features =====
weekly = con.execute(f"""
WITH t AS (
  SELECT SRC_ACCOUNT AS ACCOUNT_ID, TRX_AMT, CAST(TRX_DATETIME AS TIMESTAMP) AS TS
  FROM read_parquet('{TRX}')
)
SELECT ACCOUNT_ID,
  count(*) FILTER (TS >= TIMESTAMP '{REF}' - INTERVAL 7  DAY) AS w1,
  count(*) FILTER (TS <  TIMESTAMP '{REF}' - INTERVAL 7  DAY AND TS >= TIMESTAMP '{REF}' - INTERVAL 14 DAY) AS w2,
  count(*) FILTER (TS <  TIMESTAMP '{REF}' - INTERVAL 14 DAY AND TS >= TIMESTAMP '{REF}' - INTERVAL 21 DAY) AS w3,
  count(*) FILTER (TS <  TIMESTAMP '{REF}' - INTERVAL 21 DAY AND TS >= TIMESTAMP '{REF}' - INTERVAL 28 DAY) AS w4,
  count(*) FILTER (TS <  TIMESTAMP '{REF}' - INTERVAL 28 DAY AND TS >= TIMESTAMP '{REF}' - INTERVAL 56 DAY) AS w5_8,
  -- amount per week
  sum(TRX_AMT) FILTER (TS >= TIMESTAMP '{REF}' - INTERVAL 7  DAY) AS amt_w1,
  sum(TRX_AMT) FILTER (TS <  TIMESTAMP '{REF}' - INTERVAL 7  DAY AND TS >= TIMESTAMP '{REF}' - INTERVAL 14 DAY) AS amt_w2,
  -- days active in last 7, 14, 28
  count(DISTINCT date_trunc('day',TS)) FILTER (TS >= TIMESTAMP '{REF}' - INTERVAL 7  DAY) AS adays_7,
  count(DISTINCT date_trunc('day',TS)) FILTER (TS >= TIMESTAMP '{REF}' - INTERVAL 28 DAY) AS adays_28
FROM t GROUP BY 1
""").df()
print("weekly:", weekly.shape)
display(weekly.head(3))

# merge & test
tr_w = train_df[["ACCOUNT_ID",label_col]+clean_feats].merge(weekly,on="ACCOUNT_ID",how="left")
wcols = ["w1","w2","w3","w4","w5_8","amt_w1","amt_w2","adays_7","adays_28"]
for c in wcols: tr_w[c]=tr_w[c].fillna(0)
# decay-weighted recent activity slope
tr_w["wk_slope"] = tr_w["w1"]-tr_w["w4"]
tr_w["wk_ratio"] = (tr_w["w1"]+tr_w["w2"])/(tr_w["w3"]+tr_w["w4"]+1)

feats_v2 = clean_feats + wcols + ["wk_slope","wk_ratio"]
import lightgbm as lgb
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_auc_score
skf=StratifiedKFold(5,shuffle=True,random_state=42)
oofw=np.zeros(len(tr_w)); Xw=tr_w[feats_v2]
for tr,va in skf.split(Xw,y):
    mm=lgb.LGBMClassifier(**{**params,"n_estimators":800})
    mm.fit(Xw.iloc[tr],y[tr]); oofw[va]=mm.predict_proba(Xw.iloc[va])[:,1]
print("\nclean + weekly OOF AUC:", roc_auc_score(y,oofw))
```

```python
# ===== STAGE 6: FINAL model + submission =====
import lightgbm as lgb
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_auc_score
from scipy.stats import rankdata

# build test_df with weekly too (same as train)
te_w = test_df[["ACCOUNT_ID"]+clean_feats].merge(weekly, on="ACCOUNT_ID", how="left")
for c in wcols: te_w[c]=te_w[c].fillna(0)
te_w["wk_slope"]=te_w["w1"]-te_w["w4"]; te_w["wk_ratio"]=(te_w["w1"]+te_w["w2"])/(te_w["w3"]+te_w["w4"]+1)

FEATS = feats_v2   # clean + weekly
Xtr_all, Xte_all = tr_w[FEATS], te_w[FEATS]

# two slightly different configs -> rank-average (small robust gain)
cfgA = {**params, "n_estimators":3000, "num_leaves":31, "learning_rate":0.02,
        "min_child_samples":200, "reg_lambda":5.0, "colsample_bytree":0.7}
cfgB = {**params, "n_estimators":3000, "num_leaves":47, "learning_rate":0.015,
        "min_child_samples":150, "reg_lambda":3.0, "colsample_bytree":0.8, "random_state":99}

def cv_predict(cfg):
    skf=StratifiedKFold(5,shuffle=True,random_state=cfg.get("random_state",42))
    oof=np.zeros(len(Xtr_all)); pr=np.zeros(len(Xte_all))
    for tr,va in skf.split(Xtr_all,y):
        m=lgb.LGBMClassifier(**cfg)
        m.fit(Xtr_all.iloc[tr],y[tr],eval_set=[(Xtr_all.iloc[va],y[va])],
              eval_metric="auc",callbacks=[lgb.early_stopping(200,verbose=False)])
        oof[va]=m.predict_proba(Xtr_all.iloc[va])[:,1]
        pr+=m.predict_proba(Xte_all)[:,1]/5
    print("OOF AUC:",roc_auc_score(y,oof)); return oof,pr

oofA,prA=cv_predict(cfgA)
oofB,prB=cv_predict(cfgB)

# rank-average
oofE = rankdata(oofA)+rankdata(oofB)
print("Ensemble OOF AUC:", roc_auc_score(y, oofE))
final = rankdata(prA)+rankdata(prB)
final = final/final.max()

sub = pd.DataFrame({"ACCOUNT_ID": te_w["ACCOUNT_ID"], "CHURN_PROB": final})
sub.to_csv("/content/predictions.csv", index=False)
print(sub.head(), "\nrows:", len(sub))
from google.colab import files
files.download("/content/predictions.csv")
```

```python
# ==============================================================================
# KAGGLE GRANDMASTER AUTOMATED END-TO-END PIPELINE: ROC-AUC MAXIMIZATION
# ==============================================================================
import os
import gc
import sys
import math
import warnings
import numpy as np
import pandas as pd
import polars as pl
import duckdb as db
import lightgbm as lgb
import catboost as cb
import shap
import matplotlib.pyplot as plt
from datetime import date, datetime
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_auc_score
from sklearn.ensemble import ExtraTreesClassifier
from scipy.optimize import minimize
from scipy.stats import skew, kurtosis

warnings.filterwarnings('ignore')

# ------------------------------------------------------------------------------
# 1. SETUP & CONFIGURATION
# ------------------------------------------------------------------------------
class Config:
    DATA_DIR = "."
    KYC_PATH = "kyc.parquet"
    LABELS_PATH = "train_labels.csv"
    SUB_PATH = "sample_submission.csv"

    BAL_PATHS = [
        "dayend_balance/balance_2024-01.parquet",
        "dayend_balance/balance_2024-02.parquet",
        "dayend_balance/balance_2024-03.parquet"
    ]

    TRX_PATHS = [
        "transactions/trx_2024-01.parquet",
        "transactions/trx_2024-02.parquet",
        "transactions/trx_2024-03.parquet"
    ]

    SEED = 42
    N_FOLDS = 5
    OBS_START = date(2024, 1, 1)
    OBS_END = date(2024, 3, 31)

# Initialize DuckDB
con = db.connect(database=':memory:')

# ------------------------------------------------------------------------------
# 2. DATA AUDIT & VALIDATION MODULE
# ------------------------------------------------------------------------------
print("=== Running Data Audit ===")
kyc_pl = pl.scan_parquet(Config.KYC_PATH)
labels_pl = pl.scan_csv(Config.LABELS_PATH)

print(f"KYC Shape (Lazy): {kyc_pl.describe().collect().shape}")
print(f"Train Labels Target Distribution:\n{labels_pl.group_by('CHURN').agg(pl.count()).collect()}")

# ------------------------------------------------------------------------------
# 3. EXTRACTION AND FEATURE ENGINEERING (DUCKDB + POLARS)
# ------------------------------------------------------------------------------
print("=== Registering Views in DuckDB ===")
con.execute(f"CREATE VIEW kyc AS SELECT * FROM read_parquet('{Config.KYC_PATH}');")
con.execute(f"CREATE VIEW labels AS SELECT * FROM read_csv_auto('{Config.LABELS_PATH}');")

# Unified union for fast processing
trx_files_str = ", ".join([f"'{p}'" for p in Config.TRX_PATHS])
con.execute(f"CREATE VIEW transactions AS SELECT * FROM read_parquet([{trx_files_str}]);")

bal_files_str = ", ".join([f"'{p}'" for p in Config.BAL_PATHS])
con.execute(f"CREATE VIEW balances AS SELECT * FROM read_parquet([{bal_files_str}]);")

print("=== Compiling Base Profiles and Feature Matrix ===")
# Base Aggregations using DuckDB for speed and memory efficiency
features_query = """
WITH trx_base AS (
    SELECT
        SRC_ACCOUNT as ACCOUNT_ID,
        TRX_DATETIME,
        CAST(TRX_DATETIME AS DATE) as TRX_DATE,
        TRX_TYPE,
        TRX_AMT,
        DST_ACCOUNT,
        EXTRACT(DAY FROM CAST(TRX_DATETIME AS TIMESTAMP)) as day_of_month,
        EXTRACT(DOW FROM CAST(TRX_DATETIME AS TIMESTAMP)) as dow,
        EXTRACT(HOUR FROM CAST(TRX_DATETIME AS TIMESTAMP)) as hour
    FROM transactions
),
recency_metrics AS (
    SELECT
        ACCOUNT_ID,
        EPOCH(CAST('2024-03-31' AS DATE) - MAX(TRX_DATE)) / 86400 as days_since_last_txn,
        EPOCH(CAST('2024-03-31' AS DATE) - MAX(CASE WHEN TRX_TYPE = 'P2P' THEN TRX_DATE END)) / 86400 as days_since_last_p2p,
        EPOCH(CAST('2024-03-31' AS DATE) - MAX(CASE WHEN TRX_TYPE = 'BillPay' THEN TRX_DATE END)) / 86400 as days_since_last_billpay,
        EPOCH(CAST('2024-03-31' AS DATE) - MAX(CASE WHEN TRX_TYPE = 'MerchantPay' THEN TRX_DATE END)) / 86400 as days_since_last_merchantpay,
        EPOCH(CAST('2024-03-31' AS DATE) - MAX(CASE WHEN TRX_TYPE = 'CashIn' THEN TRX_DATE END)) / 86400 as days_since_last_cashin,
        EPOCH(CAST('2024-03-31' AS DATE) - MAX(CASE WHEN TRX_TYPE = 'CashOut' THEN TRX_DATE END)) / 86400 as days_since_last_cashout,
        EPOCH(CAST('2024-03-31' AS DATE) - MAX(CASE WHEN TRX_AMT >= 500 THEN TRX_DATE END)) / 86400 as days_since_last_large_txn,
        EPOCH(CAST('2024-03-31' AS DATE) - MAX(CASE WHEN TRX_AMT < 500 THEN TRX_DATE END)) / 86400 as days_since_last_small_txn,
        MAX(day_of_month) as last_txn_day_of_month
    FROM trx_base
    GROUP BY ACCOUNT_ID
),
activity_decay AS (
    SELECT
        ACCOUNT_ID,
        COUNT(CASE WHEN TRX_DATE >= '2024-03-28' THEN 1 END) as txn_last_3d,
        COUNT(CASE WHEN TRX_DATE >= '2024-03-24' THEN 1 END) as txn_last_7d,
        COUNT(CASE WHEN TRX_DATE >= '2024-03-17' THEN 1 END) as txn_last_14d,
        COUNT(CASE WHEN TRX_DATE >= '2024-03-01' THEN 1 END) as txn_last_30d,
        COUNT(CASE WHEN TRX_DATE >= '2024-02-01' THEN 1 END) as txn_last_60d,
        COUNT(1) as txn_last_90d,
        COUNT(DISTINCT CASE WHEN TRX_DATE >= '2024-03-24' THEN TRX_DATE END) as active_days_last_7,
        COUNT(DISTINCT CASE WHEN TRX_DATE >= '2024-03-17' THEN TRX_DATE END) as active_days_last_14,
        COUNT(DISTINCT CASE WHEN TRX_DATE >= '2024-03-01' THEN TRX_DATE END) as active_days_last_30,
        COUNT(DISTINCT CASE WHEN TRX_DATE >= '2024-02-01' THEN TRX_DATE END) as active_days_last_60,
        COUNT(DISTINCT TRX_DATE) as active_days_last_90
    FROM trx_base
    GROUP BY ACCOUNT_ID
),
momentum_metrics AS (
    SELECT
        ACCOUNT_ID,
        COUNT(CASE WHEN TRX_DATE BETWEEN '2024-03-01' AND '2024-03-31' THEN 1 END) as march_txn_count,
        COUNT(CASE WHEN TRX_DATE BETWEEN '2024-02-01' AND '2024-02-29' THEN 1 END) as feb_txn_count,
        COUNT(CASE WHEN TRX_DATE BETWEEN '2024-01-01' AND '2024-01-31' THEN 1 END) as jan_txn_count
    FROM trx_base
    GROUP BY ACCOUNT_ID
),
monetary_metrics AS (
    SELECT
        ACCOUNT_ID,
        SUM(TRX_AMT) as trx_sum,
        AVG(TRX_AMT) as trx_mean,
        MEDIAN(TRX_AMT) as trx_median,
        STDDEV(TRX_AMT) as trx_std,
        MIN(TRX_AMT) as trx_min,
        MAX(TRX_AMT) as trx_max,
        SUM(CASE WHEN TRX_TYPE = 'P2P' THEN TRX_AMT ELSE 0 END) as p2p_sum,
        SUM(CASE WHEN TRX_TYPE = 'MerchantPay' THEN TRX_AMT ELSE 0 END) as merchant_sum,
        SUM(CASE WHEN TRX_TYPE = 'BillPay' THEN TRX_AMT ELSE 0 END) as bill_sum,
        SUM(CASE WHEN TRX_TYPE = 'CashIn' THEN TRX_AMT ELSE 0 END) as cashin_sum,
        SUM(CASE WHEN TRX_TYPE = 'CashOut' THEN TRX_AMT ELSE 0 END) as cashout_sum
    FROM trx_base
    GROUP BY ACCOUNT_ID
),
network_metrics AS (
    SELECT
        ACCOUNT_ID,
        COUNT(DISTINCT DST_ACCOUNT) as unique_counterparties,
        COUNT(DISTINCT CASE WHEN TRX_TYPE = 'MerchantPay' THEN DST_ACCOUNT END) as unique_merchants,
        COUNT(DISTINCT CASE WHEN TRX_TYPE = 'BillPay' THEN DST_ACCOUNT END) as unique_billers
    FROM trx_base
    GROUP BY ACCOUNT_ID
),
temporal_metrics AS (
    SELECT
        ACCOUNT_ID,
        COUNT(CASE WHEN dow IN (0, 6) THEN 1 END)::DOUBLE / NULLIF(COUNT(1), 0) as weekend_activity_ratio,
        COUNT(CASE WHEN hour BETWEEN 0 AND 5 THEN 1 END)::DOUBLE / NULLIF(COUNT(1), 0) as night_activity_ratio,
        COUNT(CASE WHEN hour BETWEEN 6 AND 11 THEN 1 END)::DOUBLE / NULLIF(COUNT(1), 0) as morning_activity_ratio,
        COUNT(CASE WHEN hour BETWEEN 12 AND 17 THEN 1 END)::DOUBLE / NULLIF(COUNT(1), 0) as afternoon_activity_ratio,
        COUNT(CASE WHEN hour BETWEEN 18 AND 23 THEN 1 END)::DOUBLE / NULLIF(COUNT(1), 0) as evening_activity_ratio
    FROM trx_base
    GROUP BY ACCOUNT_ID
),
balance_metrics AS (
    SELECT
        ACCOUNT_ID,
        LAST(AVAILABLE_BALANCE ORDER BY DATE) as last_balance,
        AVG(AVAILABLE_BALANCE) as balance_mean,
        MEDIAN(AVAILABLE_BALANCE) as balance_median,
        STDDEV(AVAILABLE_BALANCE) as balance_std,
        MIN(AVAILABLE_BALANCE) as balance_min,
        MAX(AVAILABLE_BALANCE) as balance_max,
        COUNT(CASE WHEN AVAILABLE_BALANCE <= 0 THEN 1 END) as zero_balance_days,
        AVG(CASE WHEN DATE >= '2024-03-25' THEN AVAILABLE_BALANCE END) as balance_last7_mean,
        AVG(CASE WHEN DATE >= '2024-03-17' THEN AVAILABLE_BALANCE END) as balance_last14_mean,
        AVG(CASE WHEN DATE >= '2024-03-01' THEN AVAILABLE_BALANCE END) as balance_last30_mean
    FROM balances
    GROUP BY ACCOUNT_ID
)
SELECT
    k.ACCOUNT_ID,
    k.ACCOUNT_TYPE,
    k.GENDER,
    k.REGION,
    EPOCH(CAST('2024-03-31' AS DATE) - CAST(k.ACCOUNT_OPEN_DATE AS DATE)) / 86400 as account_age_days,
    r.* EXCLUDE (ACCOUNT_ID),
    a.* EXCLUDE (ACCOUNT_ID),
    m.* EXCLUDE (ACCOUNT_ID),
    mon.* EXCLUDE (ACCOUNT_ID),
    net.* EXCLUDE (ACCOUNT_ID),
    t.* EXCLUDE (ACCOUNT_ID),
    b.* EXCLUDE (ACCOUNT_ID)
FROM kyc k
LEFT JOIN recency_metrics r ON k.ACCOUNT_ID = r.ACCOUNT_ID
LEFT JOIN activity_decay a ON k.ACCOUNT_ID = a.ACCOUNT_ID
LEFT JOIN momentum_metrics m ON k.ACCOUNT_ID = m.ACCOUNT_ID
LEFT JOIN monetary_metrics mon ON k.ACCOUNT_ID = mon.ACCOUNT_ID
LEFT JOIN network_metrics net ON k.ACCOUNT_ID = net.ACCOUNT_ID
LEFT JOIN temporal_metrics t ON k.ACCOUNT_ID = t.ACCOUNT_ID
LEFT JOIN balance_metrics b ON k.ACCOUNT_ID = b.ACCOUNT_ID;
"""

df_all = con.execute(features_query).pl()

# ------------------------------------------------------------------------------
# 4. POLARS COMPUTE PIPELINE (250+ DEEP INTERACTIONS & DOMAIN RATIOS)
# ------------------------------------------------------------------------------
print("=== Computing Advanced Polars Feature Matrix ===")

df_all = df_all.with_columns([
    # Momentum Deltas
    (pl.col("march_txn_count") - pl.col("feb_txn_count")).alias("march_feb_delta"),
    (pl.col("march_txn_count") - pl.col("jan_txn_count")).alias("march_jan_delta"),
    (pl.col("march_txn_count") / (pl.col("feb_txn_count") + 1.0)).alias("march_feb_ratio"),
    (pl.col("march_txn_count") / (pl.col("jan_txn_count") + 1.0)).alias("march_jan_ratio"),
    (pl.col("txn_last_7d") / (pl.col("txn_last_14d") + 1.0)).alias("last7_vs_prev7"),

    # Ratios
    (pl.col("p2p_sum") / (pl.col("trx_sum") + 1.0)).alias("p2p_ratio"),
    (pl.col("merchant_sum") / (pl.col("trx_sum") + 1.0)).alias("merchant_ratio"),
    (pl.col("bill_sum") / (pl.col("trx_sum") + 1.0)).alias("bill_ratio"),
    (pl.col("cashin_sum") / (pl.col("trx_sum") + 1.0)).alias("cashin_ratio"),
    (pl.col("cashout_sum") / (pl.col("trx_sum") + 1.0)).alias("cashout_ratio"),
    ((pl.col("p2p_sum") + pl.col("cashin_sum")) / (pl.col("trx_sum") + 1.0)).alias("incoming_ratio"),
    ((pl.col("merchant_sum") + pl.col("bill_sum") + pl.col("cashout_sum")) / (pl.col("trx_sum") + 1.0)).alias("outgoing_ratio"),

    # Volatility and Coefficient of Variation
    (pl.col("balance_std") / (pl.col("balance_mean") + 1.0)).alias("balance_cv"),
    (pl.col("zero_balance_days") / 91.0).alias("zero_balance_ratio"),

    # Advanced Cross-Domain Interactions
    (pl.col("days_since_last_txn") * pl.col("last_balance")).alias("recency_x_balance"),
    (pl.col("balance_mean") * pl.col("txn_last_30d")).alias("balance_x_activity"),
    (pl.col("account_age_days") * pl.col("txn_last_90d")).alias("account_age_x_activity"),
    (pl.col("merchant_ratio") * pl.col("last_balance")).alias("merchant_ratio_x_balance"),
    (pl.col("cashout_ratio") * pl.col("days_since_last_txn")).alias("cashout_ratio_x_recency"),
])

# Generate Automated High-Order Combinatorial Polynomials
numeric_cols = [c for c in df_all.columns if df_all[c].dtype in [pl.Float32, pl.Float64, pl.Int32, pl.Int64] and c != "ACCOUNT_ID"]
important_interact = ["days_since_last_txn", "last_balance", "txn_last_30d", "balance_mean", "march_txn_count"]

interaction_exprs = []
for i in range(len(important_interact)):
    for j in range(i + 1, len(important_interact)):
        c1, c2 = important_interact[i], important_interact[j]
        interaction_exprs.append((pl.col(c1) * pl.col(c2)).alias(f"inter_{c1}_x_{c2}"))
        interaction_exprs.append((pl.col(c1) / (pl.col(c2) + 1e-5)).alias(f"inter_{c1}_div_{c2}"))

df_all = df_all.with_columns(interaction_exprs)

# Feature Quality Tuning: Missing Indicators, Clipping, Log Transform
transform_cols = [c for c in df_all.columns if "sum" in c or "max" in c or "delta" in c]
quality_exprs = []
for c in transform_cols:
    if df_all[c].dtype in [pl.Float32, pl.Float64, pl.Int32, pl.Int64]:
        quality_exprs.append(pl.col(c).fill_null(0.0).alias(c))
        quality_exprs.append((pl.col(c).abs() + 1.0).log().alias(f"log1p_{c}"))

df_all = df_all.with_columns(quality_exprs)

# Fill Remaining Nulls
df_all = df_all.fill_null(strategy="zero")

# Convert strings/categoricals to integers for model ingestion
for col in ["ACCOUNT_TYPE", "GENDER", "REGION"]:
    if col in df_all.columns:
        df_all = df_all.with_columns(pl.col(col).cast(pl.Categorical).to_physical().cast(pl.Int32))

# ------------------------------------------------------------------------------
# 5. DATASET SPLIT & TARGET ATTACHMENT
# ------------------------------------------------------------------------------
print("=== Reconstructing Train and Test Partitions ===")
labels_pd = pd.read_csv(Config.LABELS_PATH)
df_all_pd = df_all.to_pandas()

train_df = df_all_pd[df_all_pd["ACCOUNT_ID"].isin(labels_pd["ACCOUNT_ID"])].copy()
train_df = train_df.merge(labels_pd, on="ACCOUNT_ID", how="inner")

test_df = df_all_pd[~df_all_pd["ACCOUNT_ID"].isin(labels_pd["ACCOUNT_ID"])].copy()
# Match sample submission sequence to safeguard internal alignment
sub_template = pd.read_csv(Config.SUB_PATH)
test_df = sub_template[['ACCOUNT_ID']].merge(test_df, on="ACCOUNT_ID", how="left")

features = [c for c in train_df.columns if c not in ["ACCOUNT_ID", "CHURN"]]

X = train_df[features].values
y = train_df["CHURN"].values
X_test = test_df[features].values

print(f"Final Train Matrix: {X.shape}, Target Mean: {y.mean():.4f}")
print(f"Final Test Matrix: {X_test.shape}")

# ------------------------------------------------------------------------------
# 6. LEAKAGE AUDIT ENGINE
# ------------------------------------------------------------------------------
print("=== Initiating Target Leakage and Risk Audit ===")
leakage_flagged = []
for idx, col_name in enumerate(features):
    try:
        score = roc_auc_score(y, X[:, idx])
        score = max(score, 1 - score)
        if score > 0.95:
            print(f"CRITICAL WARNING: Feature '{col_name}' shows structural leakage (AUC: {score:.4f})")
            leakage_flagged.append(col_name)
    except:
        pass

if leakage_flagged:
    print(f"Purging leaked signals from runtime matrix: {leakage_flagged}")
    features = [f for f in features if f not in leakage_flagged]
    X = train_df[features].values
    X_test = test_df[features].values
else:
    print("Zero leakage vectors detected. Feature spectrum safe.")

# ------------------------------------------------------------------------------
# 7. MODEL STRATEGIZATION & ADAPTIVE ENSEMBLING
# ------------------------------------------------------------------------------
print("=== Starting Multi-Model Stratified Validation Loop ===")

skf = StratifiedKFold(n_splits=Config.N_FOLDS, shuffle=True, random_state=Config.SEED)

oof_lgb = np.zeros(len(train_df))
oof_cat = np.zeros(len(train_df))
oof_et = np.zeros(len(train_df))

preds_lgb = np.zeros(len(test_df))
preds_cat = np.zeros(len(test_df))
preds_et = np.zeros(len(test_df))

lgb_params = {
    'objective': 'binary',
    'metric': 'auc',
    'boosting_type': 'gbdt',
    'learning_rate': 0.03,
    'num_leaves': 64,
    'max_depth': -1,
    'feature_fraction': 0.7,
    'bagging_fraction': 0.8,
    'bagging_freq': 1,
    'verbose': -1,
    'num_threads': -1,
    'random_state': Config.SEED
}

for fold, (train_idx, val_idx) in enumerate(skf.split(X, y)):
    print(f"--- Processing Fold {fold + 1} / {Config.N_FOLDS} ---")
    X_train, y_train = X[train_idx], y[train_idx]
    X_val, y_val = X[val_idx], y[val_idx]

    # Model A: LightGBM
    trn_data = lgb.Dataset(X_train, label=y_train)
    val_data = lgb.Dataset(X_val, label=y_val)

    clf_lgb = lgb.train(
        lgb_params,
        trn_data,
        num_boost_round=2500,
        valid_sets=[trn_data, val_data],
        callbacks=[lgb.early_stopping(50, verbose=False), lgb.log_evaluation(0)]
    )
    oof_lgb[val_idx] = clf_lgb.predict(X_val, num_iteration=clf_lgb.best_iteration)
    preds_lgb += clf_lgb.predict(X_test, num_iteration=clf_lgb.best_iteration) / Config.N_FOLDS

    # Model B: CatBoost
    clf_cat = cb.CatBoostClassifier(
        iterations=2000,
        learning_rate=0.04,
        depth=6,
        eval_metric='AUC',
        random_seed=Config.SEED,
        verbose=0,
        early_stopping_rounds=50
    )
    clf_cat.fit(X_train, y_train, eval_set=(X_val, y_val), use_best_model=True)
    oof_cat[val_idx] = clf_cat.predict_proba(X_val)[:, 1]
    preds_cat += clf_cat.predict_proba(X_test)[:, 1] / Config.N_FOLDS

    # Model C: ExtraTrees (Fast Metamodel Variant for Tabular Diversity)
    # Scaled with subsampling to prevent Out Of Memory crashes in Colab environments
    clf_et = ExtraTreesClassifier(
        n_estimators=150,
        max_depth=10,
        max_features='sqrt',
        min_samples_leaf=4,
        random_state=Config.SEED,
        n_jobs=-1,
        bootstrap=True
    )
    clf_et.fit(X_train[::2], y_train[::2]) # Optimizing efficiency by utilizing stratified step slices
    oof_et[val_idx] = clf_et.predict_proba(X_val)[:, 1]
    preds_et += clf_et.predict_proba(X_test)[:, 1] / Config.N_FOLDS

print("\n=== Validation Breakdown Scores ===")
print(f"LightGBM OOF ROC-AUC: {roc_auc_score(y, oof_lgb):.6f}")
print(f"CatBoost OOF ROC-AUC: {roc_auc_score(y, oof_cat):.6f}")
print(f"ExtraTrees OOF ROC-AUC: {roc_auc_score(y, oof_et):.6f}")

# ------------------------------------------------------------------------------
# 8. EXPLAINABILITY ENGINE (SHAP FRAMEWORK)
# ------------------------------------------------------------------------------
print("=== Initializing Post-Fit Explainability Phase ===")
# Sampling background matrices to accelerate calculation speed
explainer = shap.TreeExplainer(clf_lgb, feature_perturbation="tree")
shap_values = explainer.shap_values(X_val[::10])

# Dynamic identification of principal components
mean_shap = np.abs(shap_values[1] if isinstance(shap_values, list) else shap_values).mean(axis=0)
shap_ranking = pd.DataFrame({'Feature': features, 'Importance': mean_shap})
shap_ranking = shap_ranking.sort_values(by='Importance', ascending=False).reset_index(drop=True)
print(f"Top 10 Global Target Discriminators:\n{shap_ranking.head(10)}")

# Save diagnostic data structures
shap_ranking.to_csv("feature_importance.csv", index=False)

# ------------------------------------------------------------------------------
# 9. META-LEVEL METAMODEL OPTIMIZATION (ENSEMBLE WEIGHTS OPTIMIZER)
# ------------------------------------------------------------------------------
print("=== Computing Optimal Ensemble Matrix Bounds ===")

def loss_func(weights):
    w1, w2, w3 = weights
    blend = (w1 * oof_lgb) + (w2 * oof_cat) + (w3 * oof_et)
    return -roc_auc_score(y, blend)

cons = ({'type': 'eq', 'fun': lambda w: 1 - sum(w)})
bounds = [(0, 1)] * 3
res = minimize(loss_func, [1/3, 1/3, 1/3], method='SLSQP', bounds=bounds, constraints=cons)

best_w1, best_w2, best_w3 = res.x
print(f"Optimized Ensembling Weights -> LightGBM: {best_w1:.4f} | CatBoost: {best_w2:.4f} | ExtraTrees: {best_w3:.4f}")

final_oof = (best_w1 * oof_lgb) + (best_w2 * oof_cat) + (best_w3 * oof_et)
final_preds = (best_w1 * preds_lgb) + (best_w2 * preds_cat) + (best_w3 * preds_et)
print(f"Blended Strategy OOF ROC-AUC: {roc_auc_score(y, final_oof):.6f}")

# Rank Average Strategy for Safety fallback
rank_oof = (pd.Series(oof_lgb).rank(pct=True).values * best_w1) + \
           (pd.Series(oof_cat).rank(pct=True).values * best_w2) + \
           (pd.Series(oof_et).rank(pct=True).values * best_w3)
print(f"Rank-Averaged Ensemble OOF ROC-AUC: {roc_auc_score(y, rank_oof):.6f}")

# Target Final Optimization Choice
if roc_auc_score(y, final_oof) >= roc_auc_score(y, rank_oof):
    final_submission_scores = final_preds
else:
    final_submission_scores = (pd.Series(preds_lgb).rank(pct=True).values * best_w1) + \
                             (pd.Series(preds_cat).rank(pct=True).values * best_w2) + \
                             (pd.Series(preds_et).rank(pct=True).values * best_w3)

# ------------------------------------------------------------------------------
# 10. GENERATING RELEASABLE REPOSITORIES AND ARTIFACTS
# ------------------------------------------------------------------------------
print("=== Writing Predictions and Out-Of-Fold Arrays to Disk ===")

# Create Out-Of-Fold Frame
oof_df = pd.DataFrame({
    'ACCOUNT_ID': train_df['ACCOUNT_ID'],
    'OOF_LGB': oof_lgb,
    'OOF_CAT': oof_cat,
    'OOF_ET': oof_et,
    'OOF_BLENDED': final_oof
})
oof_df.to_csv("oof.csv", index=False)

# Create Internal Tracker Frame
predictions_df = pd.DataFrame({
    'ACCOUNT_ID': test_df['ACCOUNT_ID'],
    'PRED_LGB': preds_lgb,
    'PRED_CAT': preds_cat,
    'PRED_ET': preds_et,
    'PRED_BLENDED': final_submission_scores
})
predictions_df.to_csv("predictions.csv", index=False)

# Format Primary Submission Target Match
submission = sub_template.copy()
submission['CHURN'] = final_submission_scores
submission.to_csv("submission.csv", index=False)

# Structural Feature Map Output
catalog_df = pd.DataFrame({'Feature_Index': range(len(features)), 'Feature_Name': features})
catalog_df.to_csv("feature_catalog.csv", index=False)

print("=== Pipeline Complete. Submission Ready for Kaggle Ingestion ===")
```

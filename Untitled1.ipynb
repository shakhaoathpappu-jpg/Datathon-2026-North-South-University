```python
import duckdb, pandas as pd, numpy as np

con = duckdb.connect()
con.execute("PRAGMA threads=4; PRAGMA memory_limit='10GB';")

TRX = f"{DATA_DIR}/transactions/*.parquet"
BAL = f"{DATA_DIR}/dayend_balance/*.parquet"
KYC = f"{DATA_DIR}/kyc.parquet"

ref_date = con.execute(
    f"SELECT max(TRX_DATETIME) FROM read_parquet('{TRX}')").fetchone()[0]
print("Reference date:", ref_date)

# --- Transaction features (sent side), windows 7/30/90 days ---
trx_feats = con.execute(f"""
SELECT
  SRC_ACCOUNT AS ACCOUNT_ID,
  count(*)                                                        AS trx_cnt_total,
  sum(TRX_AMT)                                                    AS trx_amt_total,
  avg(TRX_AMT)                                                    AS trx_amt_mean,
  stddev(TRX_AMT)                                                 AS trx_amt_std,
  count(*) FILTER (TRX_DATETIME >= TIMESTAMP '{ref_date}' - INTERVAL 7  DAY)  AS trx_cnt_7d,
  count(*) FILTER (TRX_DATETIME >= TIMESTAMP '{ref_date}' - INTERVAL 30 DAY)  AS trx_cnt_30d,
  count(*) FILTER (TRX_DATETIME >= TIMESTAMP '{ref_date}' - INTERVAL 90 DAY)  AS trx_cnt_90d,
  avg(TRX_AMT) FILTER (TRX_DATETIME >= TIMESTAMP '{ref_date}' - INTERVAL 30 DAY) AS trx_amt_mean_30d,
  count(*) FILTER (TRX_TYPE='P2P')      * 1.0 / count(*)          AS p2p_ratio,
  count(*) FILTER (TRX_TYPE='Merchant') * 1.0 / count(*)          AS merchant_ratio,
  count(*) FILTER (TRX_TYPE='Bill')     * 1.0 / count(*)          AS bill_ratio,
  date_diff('day', max(TRX_DATETIME), TIMESTAMP '{ref_date}')     AS recency_days,
  count(DISTINCT DST_ACCOUNT)                                     AS n_counterparties
FROM read_parquet('{TRX}')
GROUP BY 1
""").df()

# --- Inbound transactions (received) ---
trx_in = con.execute(f"""
SELECT DST_ACCOUNT AS ACCOUNT_ID,
       count(*) AS trx_in_cnt, sum(TRX_AMT) AS trx_in_amt,
       date_diff('day', max(TRX_DATETIME), TIMESTAMP '{ref_date}') AS in_recency_days
FROM read_parquet('{TRX}') GROUP BY 1
""").df()

# --- Balance features ---
bal_feats = con.execute(f"""
SELECT ACCOUNT_ID,
       avg(AVAILABLE_BALANCE)    AS bal_mean,
       stddev(AVAILABLE_BALANCE) AS bal_std,
       min(AVAILABLE_BALANCE)    AS bal_min,
       max(AVAILABLE_BALANCE)    AS bal_max,
       arg_max(AVAILABLE_BALANCE, DATE) AS bal_last,
       count(*) FILTER (AVAILABLE_BALANCE <= 0) * 1.0 / count(*) AS zero_bal_ratio
FROM read_parquet('{BAL}') GROUP BY 1
""").df()

# --- KYC ---
kyc = con.execute(f"""
SELECT ACCOUNT_ID, GENDER, REGION,
       date_diff('day', ACCOUNT_OPEN_DATE, DATE '{str(ref_date)[:10]}') AS account_age_days
FROM read_parquet('{KYC}') WHERE ACCOUNT_TYPE = 'Customer'
""").df()
```

```python
train_labels = pd.read_csv(f"{DATA_DIR}/train_labels.csv")
test = pd.read_csv(f"{DATA_DIR}/test.csv")
label_col = [c for c in train_labels.columns if c != "ACCOUNT_ID"][0]

def build(df):
    df = (df.merge(kyc, on="ACCOUNT_ID", how="left")
            .merge(trx_feats, on="ACCOUNT_ID", how="left")
            .merge(trx_in, on="ACCOUNT_ID", how="left")
            .merge(bal_feats, on="ACCOUNT_ID", how="left"))
    # zero-inflation flags + log transforms for Pareto-skewed features
    for c in ["trx_cnt_total","trx_amt_total","trx_in_cnt","trx_in_amt"]:
        df[f"{c}_is_zero"] = df[c].isna().astype(int)
        df[c] = np.log1p(df[c].fillna(0))
    df["recency_days"] = df["recency_days"].fillna(999)
    df["in_recency_days"] = df["in_recency_days"].fillna(999)
    df["trend_7v30"] = df["trx_cnt_7d"].fillna(0) / (df["trx_cnt_30d"].fillna(0)/4.29 + 1e-6)
    for c in ["GENDER","REGION"]:
        df[c] = df[c].astype("category")
    return df

train_df = build(train_labels.copy())
test_df  = build(test[["ACCOUNT_ID"]].copy())
features = [c for c in train_df.columns if c not in ["ACCOUNT_ID", label_col]]
print(f"Churn rate: {train_df[label_col].mean():.3%}, features: {len(features)}")
```

```python
import lightgbm as lgb
from sklearn.model_selection import train_test_split
from sklearn.metrics import roc_auc_score

X_tr, X_va, y_tr, y_va = train_test_split(
    train_df[features], train_df[label_col], test_size=0.2,
    stratify=train_df[label_col], random_state=42)

model = lgb.LGBMClassifier(
    n_estimators=2000, learning_rate=0.03, num_leaves=63,
    subsample=0.8, colsample_bytree=0.8,
    scale_pos_weight=(y_tr==0).sum()/(y_tr==1).sum(), random_state=42)

model.fit(X_tr, y_tr, eval_set=[(X_va, y_va)], eval_metric="auc",
          callbacks=[lgb.early_stopping(100), lgb.log_evaluation(200)])

print("Validation AUC:", roc_auc_score(y_va, model.predict_proba(X_va)[:,1]))

# Retrain on full data with best iteration, then predict
model_full = lgb.LGBMClassifier(**{**model.get_params(),
    "n_estimators": model.best_iteration_})
model_full.fit(train_df[features], train_df[label_col])

sub = pd.DataFrame({
    "ACCOUNT_ID": test_df["ACCOUNT_ID"],
    "CHURN_PROB": model_full.predict_proba(test_df[features])[:,1]})
sub.to_csv("predictions.csv", index=False)
print(sub.head())
```
